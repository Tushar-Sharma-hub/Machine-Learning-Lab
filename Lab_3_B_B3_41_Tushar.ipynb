```python
X = df.iloc[:, :-1].values
y = df.iloc[:, -1].values
```

```python
import numpy as np

def find_s(X, y):
    hypothesis = None
    print("Initial Hypothesis:", hypothesis)
    
    for i in range(len(X)):
        if y[i] != "unacceptable":
            if hypothesis is None:
                hypothesis = X[i].copy()
            else:
                for j in range(len(hypothesis)):
                    if hypothesis[j] != X[i][j]:
                        hypothesis[j] = "?"
            print(f"\nAfter Training Example {i + 1}")
            print(hypothesis)
            
    return hypothesis

final_hypothesis = find_s(X, y)
print("\nFinal Hypothesis:", final_hypothesis)
```

Candidate Elimination Algorithm

```python
import numpy as np

def candidate_elimination(concepts, target):
    specific_h = None
    for i in range(len(target)):
        if target[i] != "unacceptable":
            specific_h = concepts[i].copy()
            break            
    general_h = [["?" for _ in range(len(specific_h))]]   
    print("Initial S:", specific_h)
    print("Initial G:", general_h)    
    for i, h in enumerate(concepts):
        if target[i] != "unacceptable":
            general_h = [g for g in general_h if all(g[k] == '?' or g[k] == h[k] for k in range(len(specific_h)))]
            for x in range(len(specific_h)):
                if h[x] != specific_h[x]:
                    specific_h[x] = '?'
        else: 
            new_G = []
            for g in general_h:
                if all(g[k] == '?' or g[k] == h[k] for k in range(len(specific_h))):
                    for x in range(len(specific_h)):
                        if h[x] != specific_h[x]:
                            g_new = g.copy()
                            g_new[x] = specific_h[x]
                            if g_new not in new_G:
                                new_G.append(g_new)
                else:
                    new_G.append(g)
            general_h = new_G
        print(f"\nExample {i + 1}")
        print("S =", specific_h)
        print("G =", general_h)      
    return specific_h, general_h
S, G = candidate_elimination(X, y)
print("\nFinal Specific Boundary:")
print(S)
print("\nFinal General Boundary:")
for g in G:
    print(g)
```
